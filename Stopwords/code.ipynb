```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download the necessary data packages
nltk.download('stopwords')
nltk.download('punkt')

def remove_stopwords(text):
    # 1. Get the list of English stopwords
    stop_words = set(stopwords.words('english'))
    
    # 2. Tokenize the sentence (break it into individual words)
    words = word_tokenize(text)
    
    # 3. Filter out the stopwords
    # We convert word to lowercase to match the stopword list
    filtered_sentence = [w for w in words if w.lower() not in stop_words]
    
    return filtered_sentence

# Example usage
example_text = "The quick brown fox is jumping over a lazy dog."
result = remove_stopwords(example_text)

print(f"Original: {example_text}")
print(f"Filtered: {result}")
```

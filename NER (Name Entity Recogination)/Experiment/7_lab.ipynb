```python
import nltk
from nltk import word_tokenize, pos_tag, ne_chunk

def perform_ner(text):
    # 1. Tokenization: Split text into individual words
    tokens = word_tokenize(text)
    
    # 2. POS Tagging: Identify parts of speech (Nouns, Verbs, etc.)
    pos_tags = pos_tag(tokens)
    

    chunks = ne_chunk(pos_tags, binary=False)
    
    # 4. Parse the tree to extract entities
    entities = []
    for chunk in chunks:
        if hasattr(chunk, 'label'):
            entity_name = ' '.join(c[0] for c in chunk)
            entity_type = chunk.label()
            entities.append((entity_name, entity_type))
            
    return entities

# Example Usage
raw_text = "Sundar Pichai is the CEO of Google, which is headquartered in Mountain View, California."
result = perform_ner(raw_text)

print(f"{'Entity':<20} | {'Type'}")
print("-" * 35)
for entity, type in result:
    print(f"{entity:<20} | {type}")
```
